```python
#write a program to update the second line of the file
f=open("sooper.txt","r")
s=f.readlines()
s[1]="Hi how are you \n"
f.close()
f=open("sooper.txt","w")
s=f.writelines(s)
f.close()
```

```python
#program to search a particular word in a file
f = open("sooper.txt", "r")
s = f.read()
if "wood" in s:
    print("Found")
else:
    print("Not Found")
f.close()
```

```python
f = open("sooper.txt", "r")
s = f.read()
index = s.find("wood")
if index != -1:
    print("Found at index:", index)
else:
    print("Not Found")
f.close()
```

```python
#find the number of letters,digit and spaces in a file
f=open("sooper.txt","r")
n=f.read()
l=0
d=0
s=0
for i in n:
    if i.isalpha():
        l+=1
    elif i.isdigit():
        d+=1
    elif i.isspace():
        s+=1
print("number of letters : ",l)
print("number of digits : ",d)
print("number of spaces: ",s)
```
