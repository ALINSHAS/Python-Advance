```python
#find the number of letters,digits and spaces inside the file.
f=open("sooper.txt","r")
l=0
d=0
s=0
r=f.read()
for i in r:
    if i.isalpha():
        l+=1
    elif i.isdigit():
        d+=1
    elif i.isspace():
        s+=1
print("letter count",l)
print("digit count",d)
print("space count",s)
```

```python
#reverse the lines in a file
f=open("sooper.txt","r")
r=f.readlines()
f.close()
rev=r[::-1]
f=open("sooper.txt","w")
f.writelines(rev)
f.close()
```

```python
# A file totalstudents.txt contains the names of all students in a class,
# and a file passedstudents.txt contains the names of students who passed the exam.
# Write a Python program to:
# Read the names from both files.
# Find the students who did not pass.
# Write their names to a new file named failed_students.txt, one name per line.
f=open("totalstudents.txt","r")
t=f.read().split()
f.close()
f=open("passedstudents.txt","r")
p=f.read().split()
f.close()
f=open("failedstudents.txt","w")
for students in t:
    if students not in p:
        f.write(students + "\n")
f.close()
```

```python
# Two files, swiggy.txt and zomato.txt, contain the names of food items ordered from each platform.
# Write a Python program to read words from swiggy.txt and zomato.txt, combine them, and count how many times each word appears. Store the result in a dictionary and print it
f=open("swiggy.txt","r")
s=f.read().split()
f.close()
f=open("zomato.txt","r")
z=f.read().split()
f.close()
item=s+z
d={}
for i in item:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
print(d)
```

```python
# A file marks.txt contains the names and marks of students as shown below:
# Anu 80
# Rahul 65
# Meera 90
# Arun 70
# Write a Python program to read the file and write the names of students who scored more than 75 marks into a file named names.txt.
f=open("marks.txt","r")
m=f.readlines()
f.close()
f=open("names.txt","w")
for i in m:
    name,marks = i.split()
    if int(marks) > 75:
        f.write(name + "\n")
f.close()
```

```python
def fileappend(file_name,file_name2):
    file_name=input("Enter the file name : ")
    file_name2=input("Enter the file to which it is copied")
    f = open(file_name, "a")
    r = f.read()
    f.close()
    f =open(file_name2,"a")
    f.write(r)
    f.close()
fileappend(file_name,file_name2)
```
